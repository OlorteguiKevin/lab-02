```python
from pycaret.classification import *
import warnings
warnings.filterwarnings("ignore")
```

```python
s = setup(data=data, target='Outcome', session_id=123)

```

```python

# Comparar todos los modelos
best = compare_models()

# Mostrar las métricas del mejor modelo
evaluate_model(best)
```

```python
# Optimizar el mejor modelo encontrado
tuned_model = tune_model(best, optimize='Accuracy')
```

```python
# Realizar predicciones en los datos originales
predictions = predict_model(tuned_model)
print(predictions[['Outcome', 'prediction_label']].head())
```

```python
# Guardar el modelo optimizado
save_model(tuned_model, 'modelo_diabetes_final')
```
